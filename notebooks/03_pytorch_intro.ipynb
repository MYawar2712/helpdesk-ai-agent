# Day 10 — PyTorch foundations

A small tensor, autograd, and training-loop exercise. Run this notebook after installing PyTorch.

```python
import torch
from torch import nn

torch.manual_seed(42)
x = torch.randn(128, 2)
y = ((x[:, 0] + x[:, 1]) > 0).long()
print(x.shape, y.shape)
```

```python
model = nn.Sequential(nn.Linear(2, 8), nn.ReLU(), nn.Linear(8, 2))
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

for _epoch in range(100):
    optimizer.zero_grad()
    loss = loss_fn(model(x), y)
    loss.backward()
    optimizer.step()

print(f"final loss: {loss.item():.4f}")
```

```python
with torch.no_grad():
    predictions = model(x).argmax(dim=1)
accuracy = (predictions == y).float().mean()
print(f"accuracy: {accuracy.item():.3f}")
```

## Concepts

Tensors hold data, `requires_grad` enables autograd, `backward()` computes gradients, and the optimizer updates model parameters.
